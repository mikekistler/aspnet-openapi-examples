## Types of errors

There are several types of errors that can occur in an ASP.NET Minimal API app:

- **Exceptions**: These are errors that occur during the execution of the app. They can be handled using try-catch blocks.
- **HTTP errors**: These are errors that occur when the client makes a request that the server cannot fulfill. They are represented by HTTP status codes (e.g., 404 Not Found, 500 Internal Server Error).
- **Validation errors**: These are errors that occur when the client sends invalid data to the server. They can be handled using model validation.

All these types of errors can be handled either by middleware or in the app code.

## Handling exceptions

Exceptions can occur at many points in the request processing, including in the middleware pipeline, in the route handler, or in the response formatter. Some of this code is provided by the framework, and some is written by you.

Wherever possible, you should handle exceptions in your app code using try-catch blocks.

## Handling HTTP errors

HTTP errors are represented by HTTP status codes. You can return an HTTP error from your app code by setting the status code on the response object.

Middleware may also generate HTTP errors. For example, the parameter binding middleware generates a 400 Bad Request response if a required parameter is missing.
